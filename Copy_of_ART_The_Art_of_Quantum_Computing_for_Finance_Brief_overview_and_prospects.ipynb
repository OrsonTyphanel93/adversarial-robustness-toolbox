\begin{align*}
    n &= \text{number of qubits} \\
    N &= \text{number of samples} \\
    S &= \text{shape of output array} = (m_1, m_2) \in \mathbb{N}^2
\end{align*}



 Procedure for creating quantum circuits,
\begin{align*}
    U_{QC}(n,\epsilon,\nu) : \mathcal{H}_n \rightarrow \mathcal{H}_n
\end{align*}




Distribution of probabilities over measurement results,
\begin{align*}
    P(x|\psi) = |\langle x | U_{QC}(n,\epsilon,\nu) |\psi\rangle|^2 \quad x \in \{0,1\}^n
\end{align*}




 Process of parallel circuit execution,
\begin{align*}
    P_i(x|\psi) &= |\langle x | U^{(i)}_{QC}(n,\epsilon,\nu) |\psi\rangle|^2 \\
    i &\in \{1,2,3\}
\end{align*}


The combined probability distribution,

\begin{align*}
    P_c(x) = \frac{1}{N}\sum_{i=1}^3 \sum_{j=1}^N \delta(x,x_j^{(i)})
\end{align*}


 The practice of creating noise patterns,
\begin{align*}
    \eta(m_1,m_2) = \frac{\xi(m_1m_2) - \frac{2^n-1}{2}}{\frac{2^n-1}{2}}
\end{align*}



\begin{align*}
    \eta'(m_1,m_2) = \frac{\eta(m_1,m_2) - \mathbb{E}[\eta]}{\sqrt{\text{Var}[\eta]}}
\end{align*}

```python
class QuantumDiffusionModel:
    """
    Class implementing quantum circuit synthesis using advanced diffusion models.
    Based on recent research in quantum circuit synthesis with multimodal diffusion models.

    Attributes:
        num_qubits (int): Number of qubits in the circuit
        num_layers (int): Number of layers in the circuit
        available_gates (list): List of supported quantum gates
        device (str): Device to run the model on ('cpu' or 'cuda')
        embedding_dim (int): Dimensionality of gate embeddings
    """

    def __init__(self, num_qubits: int = 8, num_layers: int = 4,
                 embedding_dim: int = 128, device: str = 'cpu'):
        self.num_qubits = num_qubits
        self.num_layers = num_layers
        self.available_gates = ['h', 'cx', 'ccx', 'swap', 'rx', 'ry', 'rz', 'cp']
        self.embedding_dim = embedding_dim
        self.device = device

        # Initialize parameters for diffusion process
        self.diffusion_steps = 50
        self.guidance_scale = 7.5

        # Set random seed for reproducibility
        torch.manual_seed(42)

        # Initialize gate embeddings based on research recommendations
        self._initialize_gate_embeddings()

    def _initialize_gate_embeddings(self):
        """Initialize gate embeddings using orthogonal matrices."""
        self.gate_embeddings = {}
        for gate in self.available_gates:
            # Create orthogonal random embeddings as recommended in research
            embedding = torch.randn(self.embedding_dim)
            embedding = embedding / embedding.norm()
            self.gate_embeddings[gate] = embedding

    def _encode_circuit(self, circuit_tensor: torch.Tensor) -> torch.Tensor:
        """
        Encode circuit tensor into three-dimensional representation as per research.

        Args:
            circuit_tensor (torch.Tensor): Circuit representation tensor

        Returns:
            torch.Tensor: Three-dimensional encoded circuit tensor
        """
        # Reshape to match research recommendations (qubit index × time step × gate)
        encoded_circuit = circuit_tensor.view(
            self.num_layers,
            self.num_qubits,
            self.embedding_dim
        )
        return encoded_circuit

    def _decode_circuit(self, encoded_circuit: torch.Tensor) -> QuantumCircuit:
        """
        Decode tensor into quantum circuit using cosine similarity matching.

        Args:
            encoded_circuit (torch.Tensor): Encoded circuit tensor

        Returns:
            QuantumCircuit: Generated quantum circuit
        """
        qr = QuantumRegister(self.num_qubits, 'q')
        cr = ClassicalRegister(self.num_qubits, 'c')
        qc = QuantumCircuit(qr, cr)

        # Iterate through layers and qubits
        for layer in range(self.num_layers):
            for qubit_idx in range(self.num_qubits):
                # Get gate embedding for current position
                current_embedding = encoded_circuit[layer, qubit_idx]

                # Find best matching gate using cosine similarity
                similarities = {
                    gate: torch.nn.functional.cosine_similarity(
                        current_embedding.unsqueeze(0),
                        embedding.unsqueeze(0)
                    ).item()
                    for gate, embedding in self.gate_embeddings.items()
                }

                # Select gate with highest similarity
                selected_gate = max(similarities.items(), key=lambda x: x[1])[0]

                # Apply selected gate
                if selected_gate.startswith('r'):
                    angle = np.pi/2  # Default rotation angle
                    getattr(qc, selected_gate)(angle, qr[qubit_idx])
                elif selected_gate == 'h':
                    qc.h(qr[qubit_idx])
                elif selected_gate == 'cx':
                    # Add entangling gates conditionally
                    if qubit_idx < self.num_qubits - 1:
                        qc.cx(qr[qubit_idx], qr[qubit_idx + 1])

        return qc

    def _generate_circuit_tensor(self, prompt_tensor: torch.Tensor) -> torch.Tensor:
        """Generate circuit tensor using multimodal diffusion pipeline."""
        # Use the Hugging Face token for authentication
        from google.colab import userdata
        HF_TOKEN = userdata.get('HF_TOKEN')

        pipe = StableDiffusionPipeline.from_pretrained(
            'runwayml/stable-diffusion-v1-5',
            device_map="balanced",
            use_auth_token=HF_TOKEN
        )

        # Generate tensor representation
        with torch.no_grad():
            outputs = pipe(
                prompt_tensor,
                num_inference_steps=self.diffusion_steps,
                guidance_scale=self.guidance_scale
            )

        # Process generated tensor to match circuit structure
        circuit_tensor = outputs.images[0]
        return self._encode_circuit(circuit_tensor)

    def generate_circuit(self, prompt: str) -> QuantumCircuit:
        """Generate quantum circuit based on text prompt."""
        # Encode prompt
        tokenizer = AutoTokenizer.from_pretrained('openai/clip-vit-base-patch32')
        inputs = tokenizer(prompt, return_tensors='pt', max_length=512, truncation=True)
        prompt_tensor = inputs['input_ids'].to(self.device)

        # Generate circuit tensor
        circuit_tensor = self._generate_circuit_tensor(prompt_tensor)

        # Decode tensor to quantum circuit
        return self._decode_circuit(circuit_tensor)

    def plot_circuit(self, circuit: QuantumCircuit):
        """Plot the generated quantum circuit."""
        import matplotlib.pyplot as plt

        # Create figure
        fig = plt.figure(figsize=(12, 6))
        ax = fig.add_subplot(111)

        # Draw circuit
        circuit.draw(output='mpl', ax=ax)

        plt.title('Generated Quantum Circuit')
        plt.tight_layout()
        plt.show()
```

```python
# @title

import numpy as np
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
import matplotlib.pyplot as plt
from typing import Dict, List, Optional, Tuple

def _integrate_diffusion_model(self,qc: QuantumCircuit, qr: QuantumRegister,
                             diffusion_model: QuantumDiffusionModel,
                             dd_sequence: str = 'XY4') -> None:
    """
    Integrates a quantum diffusion model into the quantum circuit with advanced DD sequences.

    Args:
        qc (QuantumCircuit): The quantum circuit to integrate into
        qr (QuantumRegister): The quantum register containing the qubits
        diffusion_model (QuantumDiffusionModel): The diffusion model instance
        dd_sequence (str): Dynamical decoupling sequence ('XY4' or 'CPMG')
    """
    num_qubits_needed = diffusion_model.num_qubits

    if len(qr) < num_qubits_needed:
        raise ValueError(f"Circuit requires at least {num_qubits_needed} qubits, "
                        f"but only {len(qr)} qubits are available")

    selected_qubits = qr[:num_qubits_needed]
    template_circuit = diffusion_model.generate_circuit("template")
    template_instruction = template_circuit.to_instruction()

    # Add advanced DD pulses before integration
    _apply_advanced_dd(qc, selected_qubits, dd_sequence)

    qc.append(template_instruction, selected_qubits)

    # Add error correction checks with syndrome measurement
    for i in range(num_qubits_needed):
        _add_error_check_with_syndrome(qc, qr, 'anc', i)

    # Add final DD sequence
    _apply_advanced_dd(qc, selected_qubits, dd_sequence)

def create_quantum_circuit(self,num_qubits: int = 8,
                         num_classical_bits: Optional[int] = None,
                         error_correction: bool = True,
                         noise_simulation: bool = False,
                         optimization_level: int = 3,
                         diffusion_model: Optional['QuantumDiffusionModel'] = None,
                         dd_sequence: str = 'XY4',
                         surface_code_distance: int = 3) -> QuantumCircuit:
    """
    Creates an advanced quantum circuit with enhanced error correction, noise resilience,
    and quantum diffusion model integration.

    Args:
        num_qubits (int): Number of qubits in the circuit
        num_classical_bits (Optional[int]): Number of classical bits for measurement
        error_correction (bool): Whether to include error correction
        noise_simulation (bool): Whether to simulate realistic noise
        optimization_level (int): Qiskit optimization level (0-3)
        diffusion_model (Optional[QuantumDiffusionModel]): Quantum diffusion model instance
        dd_sequence (str): Dynamical decoupling sequence ('XY4' or 'CPMG')
        surface_code_distance (int): Surface code distance parameter

    Returns:
        QuantumCircuit: Optimized quantum circuit with advanced considerations
    """
    num_classical_bits = num_qubits if num_classical_bits is None else num_classical_bits
    qr = QuantumRegister(num_qubits, 'q')
    cr = ClassicalRegister(num_classical_bits, 'c')
    ancilla = QuantumRegister(2, 'anc')

    qc = QuantumCircuit(qr, ancilla, cr)

    if noise_simulation:
        _add_dynamical_decoupling(qc, qr,dd_sequence)

    if error_correction:
        _apply_surface_code_pattern(qc, qr, surface_code_distance)

    for i in range(num_qubits):
        qc.h(qr[i])

    if error_correction:
        _add_error_check_with_syndrome(qc, qr, ancilla, i)

    for i in range(num_qubits-1):
        qc.crx(np.pi/4, qr[i], qr[num_qubits-1])

    if error_correction:
        _add_error_check_with_syndrome(qc, qr, ancilla, num_qubits-1)

    if diffusion_model:
        _integrate_diffusion_model(qc, qr, diffusion_model, dd_sequence)

    _add_measurement_error_mitigation(qc, qr, cr)
    _apply_surface_code_pattern(qc, qr, surface_code_distance)

    return transpile(qc, optimization_level=optimization_level)

def _apply_advanced_dd(self,qc: QuantumCircuit, qr: QuantumRegister,
                       sequence_type: str = 'XY4') -> None:
    """Implements advanced dynamical decoupling sequences."""
    if sequence_type == 'XY4':
        pulse_sequence = ['Y', 'X', '[-X]', '[-Y]', 'Y', 'X', '[-X]', '[-Y]']
    elif sequence_type == 'CPMG':
        pulse_sequence = ['X', 'X']
    else:
        raise ValueError("Unsupported DD sequence")

    for gate in pulse_sequence:
        for qubit in qr:
            if gate.startswith('[-'):
                qc.x(qubit).inverse()
            elif gate == 'X':
                qc.x(qubit)
            elif gate == 'Y':
                qc.y(qubit)

def _apply_surface_code_pattern(self,qc: QuantumCircuit, qr: QuantumRegister,
                               distance: int = 3) -> None:
    """Implements surface code pattern with configurable distance."""
    for i in range(0, len(qr)-1, 2):
        # Add stabilizer measurements
        qc.cz(qr[i], qr[i+1])

        if i < len(qr)-2:
            qc.cx(qr[i], qr[i+2])
            qc.cx(qr[i+1], qr[i+2])
            qc.cx(qr[i], qr[i+2])

def _add_error_check_with_syndrome(self,qc: QuantumCircuit, qr: QuantumRegister,
                                 ancilla: QuantumRegister, qubit_index: int) -> None:
    """Adds error checking with syndrome measurement and correction."""
    # Measure syndrome bits for error detection
    qc.ccx(qr[qubit_index], qr[(qubit_index+1)%len(qr)], ancilla[0])
    qc.cx(qr[qubit_index], ancilla[1])

    # Additional syndrome measurements for better error detection
    qc.ccx(qr[(qubit_index+1)%len(qr)], qr[(qubit_index+2)%len(qr)], ancilla[0])
    qc.cx(qr[(qubit_index+1)%len(qr)], ancilla[1])

    # Conditional correction based on syndrome
    qc.ccx(ancilla[0], ancilla[1], qr[(qubit_index+2)%len(qr)])

def _add_measurement_error_mitigation(self,qc: QuantumCircuit, qr: QuantumRegister,
                                    cr: ClassicalRegister) -> None:
    """Adds measurement error mitigation techniques."""
    for i in range(len(qr)):
        # Add multiple measurements for error reduction
        for _ in range(3):
            qc.measure(qr[i], cr[i])


def _add_dynamical_decoupling(self,qc: QuantumCircuit, qr: QuantumRegister, dd_sequence: str) -> None:
    """Adds dynamical decoupling pulses to reduce noise"""
    pulse_sequence = []
    if dd_sequence == 'XY4':
        pulse_sequence = ['X', 'Y', '[-X]', '[-Y]']
    elif dd_sequence == 'CPMG':
        pulse_sequence = ['X', 'X']
    # Add other sequences as needed

    for gate in pulse_sequence:
        for qubit in qr:
            if gate.startswith('[-'):
                qc.x(qubit).inverse()
            elif gate == 'X':
                qc.x(qubit)
            elif gate == 'Y':
                qc.y(qubit)


def _apply_surface_code_pattern(self,qc: QuantumCircuit, qr: QuantumRegister) -> None:
    """Implements surface code pattern for error correction"""
    for i in range(0, len(qr)-1, 2):
        qc.cz(qr[i], qr[i+1])
        if i < len(qr)-2:
            qc.cx(qr[i], qr[i+2])
            qc.cx(qr[i+1], qr[i+2])
            qc.cx(qr[i], qr[i+2])

def _add_error_check(self,qc: QuantumCircuit, qr: QuantumRegister, ancilla: QuantumRegister, qubit_index: int) -> None:
    """Adds error checking after quantum operations"""
    # Measure syndrome bits for error detection
    qc.ccx(qr[qubit_index], qr[(qubit_index+1)%len(qr)], ancilla[0])
    qc.cx(qr[qubit_index], ancilla[1])
    qc.ccx(qr[(qubit_index+1)%len(qr)], qr[(qubit_index+2)%len(qr)], ancilla[0])
    qc.cx(qr[(qubit_index+1)%len(qr)], ancilla[1])
    qc.ccx(ancilla[0], ancilla[1], qr[(qubit_index+2)%len(qr)])



def generate_quantum_perturbations(self,circuit: QuantumCircuit,
                                num_samples: int = 1000,
                                noise_model: Optional[str] = None) -> Dict[str, float]:
    """
    Generates quantum perturbations using AerSimulator with advanced noise models.

    Args:
        circuit (QuantumCircuit): Quantum circuit to simulate
        num_samples (int): Number of shots for simulation
        noise_model (Optional[str]): Noise model to apply

    Returns:
        Dict[str, float]: Probabilities of measurement outcomes
    """
    backend = AerSimulator(noise_model=noise_model)
    job = backend.run(circuit, shots=num_samples)
    counts = job.result().get_counts(circuit)
    probabilities = {k: v/num_samples for k, v in counts.items()}
    return probabilities

def plot_circuit_characteristics(self,circuit: QuantumCircuit) -> None:
    """
    Creates visual representations of quantum circuit characteristics using multiple subplots.

    Args:
        circuit (QuantumCircuit): The quantum circuit to analyze and visualize

    Returns:
        None: Displays multiple matplotlib figures showing circuit characteristics
    """
    import matplotlib.pyplot as plt
    import numpy as np
    from qiskit.visualization import plot_histogram

    # Create a figure with multiple subplots
    fig = plt.figure(figsize=(15, 10))

    # 1. Circuit Diagram
    plt.subplot(221)
    circuit.draw(output='mpl')
    plt.title('Circuit Diagram')

    # 2. Gate Distribution
    plt.subplot(222)
    gate_counts = circuit.count_ops()
    gates = list(gate_counts.keys())
    counts = list(gate_counts.values())
    plt.bar(gates, counts)
    plt.title('Gate Distribution')
    plt.xticks(rotation=45)

    # 3. Circuit Depth Analysis
    plt.subplot(223)
    depth = circuit.depth
    width = circuit.width()
    plt.bar(['Depth', 'Width'], [depth, width])
    plt.title('Circuit Metrics')

    # 4. Qubit Connectivity
    plt.subplot(224)
    num_qubits = circuit.num_qubits
    connectivity = np.zeros((num_qubits, num_qubits))

    for op, qubits, _ in circuit.data:
        if len(qubits) == 2:
            q1, q2 = qubits
            connectivity[q1.index][q2.index] = 1
            connectivity[q2.index][q1.index] = 1

    plt.imshow(connectivity, cmap='Blues')
    plt.colorbar()
    plt.title('Qubit Connectivity')

    plt.tight_layout()
    plt.show()
```

```python
# @title

import numpy as np
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
import matplotlib.pyplot as plt
from typing import Dict, List, Optional, Tuple
from qiskit_aer import AerSimulator
from qiskit import transpile

class QuantumNoiseGenerator:
    def __init__(self, num_qubits=8, error_correction=True, noise_simulation=True):
        """
        Initialize the QuantumNoiseGenerator with optimized parameters.

        Args:
            num_qubits (int): Number of qubits for noise generation
            error_correction (bool): Enable error correction in circuits
            noise_simulation (bool): Enable noise simulation
        """
        self.num_qubits = num_qubits
        self.error_correction = error_correction
        self.noise_simulation = noise_simulation

        # Initialize backend with optimized settings
        self.backend = AerSimulator(
            method='density_matrix',  # Best for noise simulation
            blocking_enable=True,  # Enable parallelization
            blocking_qubits=16,  # Optimal chunk size for most systems
            max_parallel_experiments=0,  # Enable parallel circuit execution
            max_parallel_shots=0,  # Enable parallel shot execution
            shot_branching_enable=True  # Enable shot branching for better performance
        )

        # Initialize QuantumDiffusionModel instance
        # Assuming QuantumDiffusionModel is defined elsewhere and is accessible
        self.diffusion_model_instance = QuantumDiffusionModel(num_qubits=self.num_qubits)


        # Create base circuit with optimized parameters
        self.circuit = self.create_quantum_circuit(
            num_qubits=self.num_qubits,
            error_correction=self.error_correction,
            noise_simulation=self.noise_simulation,
            optimization_level=3,  # Optimal optimization level
            dd_sequence='XY4'  # More robust DD sequence
        )
        self.circuit.measure_all()
       # self.plot_circuit_characteristics(self.circuit)
        #self.plot_circuit()


    def _integrate_diffusion_model(self,qc: QuantumCircuit, qr: QuantumRegister,
                             diffusion_model: QuantumDiffusionModel,
                             dd_sequence: str = 'XY4') -> None:
        """
        Integrates a quantum diffusion model into the quantum circuit with advanced DD sequences.

        Args:
            qc (QuantumCircuit): The quantum circuit to integrate into
            qr (QuantumRegister): The quantum register containing the qubits
            diffusion_model (QuantumDiffusionModel): The diffusion model instance
            dd_sequence (str): Dynamical decoupling sequence ('XY4' or 'CPMG')
        """
        num_qubits_needed = diffusion_model.num_qubits

        if len(qr) < num_qubits_needed:
            raise ValueError(f"Circuit requires at least {num_qubits_needed} qubits, "
                            f"but only {len(qr)} qubits are available")

        selected_qubits = qr[:num_qubits_needed]
        template_circuit = diffusion_model.generate_circuit("template")
        template_instruction = template_circuit.to_instruction()

        # Add advanced DD pulses before integration
        self._apply_advanced_dd(qc, selected_qubits, dd_sequence)

        qc.append(template_instruction, selected_qubits)

        # Add error correction checks with syndrome measurement
        for i in range(num_qubits_needed):
            self._add_error_check_with_syndrome(qc, qr, 'anc', i)

        # Add final DD sequence
        self._apply_advanced_dd(qc, selected_qubits, dd_sequence)


    def create_quantum_circuit(self,num_qubits: int = 8,
                            num_classical_bits: Optional[int] = None,
                            error_correction: bool = True,
                            noise_simulation: bool = False,
                            optimization_level: int = 3,
                            diffusion_model: Optional['QuantumDiffusionModel'] = None,
                            dd_sequence: str = 'XY4',
                            surface_code_distance: int = 3,
                            seed_transpiler: Optional[int] = None,
                            seed_simulator: Optional[int] = None) -> QuantumCircuit:
        """
        Creates an advanced quantum circuit with enhanced error correction, noise resilience,
        and quantum diffusion model integration.

        Args:
            num_qubits (int): Number of qubits in the circuit
            num_classical_bits (Optional[int]): Number of classical bits for measurement
            error_correction (bool): Whether to include error correction
            noise_simulation (bool): Whether to simulate realistic noise
            optimization_level (int): Qiskit optimization level (0-3)
            diffusion_model (Optional[QuantumDiffusionModel]): Quantum diffusion model instance
            dd_sequence (str): Dynamical decoupling sequence ('XY4' or 'CPMG')
            surface_code_distance (int): Surface code distance parameter
            seed_transpiler (Optional[int]): Seed for transpiler randomness
            seed_simulator (Optional[int]): Seed for simulator randomness

        Returns:
            QuantumCircuit: Optimized quantum circuit with advanced considerations
        """
        num_classical_bits = num_qubits if num_classical_bits is None else num_classical_bits
        qr = QuantumRegister(num_qubits, 'q')
        cr = ClassicalRegister(num_classical_bits, 'c')
        ancilla = QuantumRegister(2, 'anc')

        qc = QuantumCircuit(qr, ancilla, cr)

        if noise_simulation:
            self._add_dynamical_decoupling(qc, qr,dd_sequence)

        if error_correction:
            self._apply_surface_code_pattern(qc, qr, surface_code_distance)

        for i in range(num_qubits):
            qc.h(qr[i])

        if error_correction:
            self._add_error_check_with_syndrome(qc, qr, ancilla, i) # Call class method

        for i in range(num_qubits-1):
            # Replace crx with equivalent gates for density matrix method compatibility
            qc.cx(qr[i], qr[num_qubits-1])
            qc.rz(-np.pi/4, qr[num_qubits-1])
            qc.cx(qr[i], qr[num_qubits-1])
            qc.rz(np.pi/4, qr[num_qubits-1])


        if error_correction:
            self._add_error_check_with_syndrome(qc, qr, ancilla, num_qubits-1) # Call class method

        if diffusion_model:
            self._integrate_diffusion_model(qc, qr, diffusion_model, dd_sequence)

        self._add_measurement_error_mitigation(qc, qr, cr)
        self._apply_surface_code_pattern(qc, qr, surface_code_distance)

        return transpile(qc, optimization_level=optimization_level, seed_transpiler=seed_transpiler)

    def _apply_advanced_dd(self,qc: QuantumCircuit, qr: QuantumRegister,
                        sequence_type: str = 'XY4') -> None:
        """Implements advanced dynamical decoupling sequences."""
        if sequence_type == 'XY4':
            pulse_sequence = ['Y', 'X', '[-X]', '[-Y]', 'Y', 'X', '[-X]', '[-Y]']
        elif sequence_type == 'CPMG':
            pulse_sequence = ['X', 'X']
        else:
            raise ValueError("Unsupported DD sequence")

        for gate in pulse_sequence:
            for qubit in qr:
                if gate.startswith('[-'):
                    qc.x(qubit).inverse()
                elif gate == 'X':
                    qc.x(qubit)
                elif gate == 'Y':
                    qc.y(qubit)

    def _apply_surface_code_pattern(self,qc: QuantumCircuit, qr: QuantumRegister,
                                distance: int = 3) -> None:
        """Implements surface code pattern with configurable distance."""
        for i in range(0, len(qr)-1, 2):
            # Add stabilizer measurements
            qc.cz(qr[i], qr[i+1])

            if i < len(qr)-2:
                qc.cx(qr[i], qr[i+2])
                qc.cx(qr[i+1], qr[i+2])
                qc.cx(qr[i], qr[i+2])

    def _add_error_check_with_syndrome(self,qc: QuantumCircuit, qr: QuantumRegister,
                                    ancilla: QuantumRegister, qubit_index: int) -> None:
        """Adds error checking with syndrome measurement and correction."""
        # Measure syndrome bits for error detection
        qc.ccx(qr[qubit_index], qr[(qubit_index+1)%len(qr)], ancilla[0])
        qc.cx(qr[qubit_index], ancilla[1])

        # Additional syndrome measurements for better error detection
        qc.ccx(qr[(qubit_index+1)%len(qr)], qr[(qubit_index+2)%len(qr)], ancilla[0])
        qc.cx(qr[(qubit_index+1)%len(qr)], ancilla[1])

        # Conditional correction based on syndrome
        qc.ccx(ancilla[0], ancilla[1], qr[(qubit_index+2)%len(qr)])

    def _add_measurement_error_mitigation(self,qc: QuantumCircuit, qr: QuantumRegister,
                                        cr: ClassicalRegister) -> None:
        """Adds measurement error mitigation techniques."""
        for i in range(len(qr)):
            # Add multiple measurements for error reduction
            for _ in range(3):
                qc.measure(qr[i], cr[i])


    def _add_dynamical_decoupling(self,qc: QuantumCircuit, qr: QuantumRegister, dd_sequence: str) -> None:
        """Adds dynamical decoupling pulses to reduce noise"""
        pulse_sequence = []
        if dd_sequence == 'XY4':
            pulse_sequence = ['X', 'Y', '[-X]', '[-Y]']
        elif dd_sequence == 'CPMG':
            pulse_sequence = ['X', 'X']
        # Add other sequences as needed

        for gate in pulse_sequence:
            for qubit in qr:
                if gate.startswith('[-'):
                    qc.x(qubit).inverse()
                elif gate == 'X':
                    qc.x(qubit)
                elif gate == 'Y':
                    qc.y(qubit)


    def _apply_surface_code_pattern(self,qc: QuantumCircuit, qr: QuantumRegister, distance: int = 3) -> None:
        """Implements surface code pattern for error correction"""
        for i in range(0, len(qr)-1, 2):
            qc.cz(qr[i], qr[i+1])
            if i < len(qr)-2:
                qc.cx(qr[i], qr[i+2])
                qc.cx(qr[i+1], qr[i+2])
                qc.cx(qr[i], qr[i+2])

    def _add_error_check(self,qc: QuantumCircuit, qr: QuantumRegister, ancilla: QuantumRegister, qubit_index: int) -> None:
        """Adds error checking after quantum operations"""
        # Measure syndrome bits for error detection
        qc.ccx(qr[qubit_index], qr[(qubit_index+1)%len(qr)], ancilla[0])
        qc.cx(qr[qubit_index], ancilla[1])
        qc.ccx(qr[(qubit_index+1)%len(qr)], qr[(qubit_index+2)%len(qr)], ancilla[0])
        qc.cx(qr[(qubit_index+1)%len(qr)], ancilla[1])
        qc.ccx(ancilla[0], ancilla[1], qr[(qubit_index+2)%len(qr)])


    def generate_quantum_noise(self, shape, num_samples=1000):
        """
        Generate quantum noise patterns using optimized quantum circuit simulation.

        Args:
            shape (tuple): Shape of the noise pattern to generate
            num_samples (int): Number of measurement shots per circuit

        Returns:
            numpy.ndarray: Generated noise pattern
        """
        # Configure backend for optimal performance
        self.backend.set_options(
            max_memory_mb=0,  # Use system memory size
            statevector_parallel_threshold=16,  # Enable parallelization for larger circuits
            batched_shots_gpu=True  # Enable batched execution on GPU
        )

        # Create multiple circuits with different seeds for diverse noise patterns
        circuits = [
            self.create_quantum_circuit(
                num_qubits=self.num_qubits,
                error_correction=self.error_correction,
                noise_simulation=self.noise_simulation,
                seed_transpiler=i,  # Different seeds for diversity
                seed_simulator=i  # Different seeds for reproducibility
            )
            for i in range(3)
        ]

        # Run all circuits in parallel with optimized batch size
        batch_size = 1024  # Optimal batch size for most systems
        jobs = []
        for circuit in circuits:
            # Split shots into batches for better memory management
            num_batches = (num_samples + batch_size - 1) // batch_size
            for batch in range(num_batches):
                shots = min(batch_size, num_samples - batch * batch_size)
                job = self.backend.run(
                    circuit,
                    shots=shots,
                    seed_simulator=batch  # Different seeds for batches
                )
                jobs.append(job)

        # Collect and combine results efficiently
        results = []
        for job in jobs:
            result = job.result()
            counts = result.get_counts()
            results.append(counts)

        # Combine probabilities from all circuits
        combined_probs = {}
        for result in results:
            for key, value in result.items():
                combined_probs[key] = combined_probs.get(key, 0) + value

        # Normalize probabilities
        total_shots = sum(combined_probs.values())
        combined_probs = {k: v/total_shots for k, v in combined_probs.items()}

        # Generate noise pattern based on quantum measurements
        flat_shape = shape[0] * shape[1]
        noise_pattern = np.random.choice(
            list(combined_probs.keys()),
            size=flat_shape,
            p=list(combined_probs.values())
        ).reshape(shape)

        # Scale noise appropriately
        max_bits = len(list(combined_probs.keys())[0])
        scaled_noise = (noise_pattern.astype(float) / (2**max_bits - 1) - 0.5)

        return scaled_noise

    def generate_quantum_perturbations(self,circuit: QuantumCircuit,
                                num_samples: int = 1000,
                                noise_model: Optional[str] = None) -> Dict[str, float]:
        """
        Generates quantum perturbations using AerSimulator with advanced noise models.

        Args:
            circuit (QuantumCircuit): Quantum circuit to simulate
            num_samples (int): Number of shots for simulation
            noise_model (Optional[str]): Noise model to apply

        Returns:
            Dict[str, float]: Probabilities of measurement outcomes
        """
        backend = AerSimulator(noise_model=noise_model)
        job = backend.run(circuit, shots=num_samples)
        counts = job.result().get_counts(circuit)
        probabilities = {k: v/num_samples for k, v in counts.items()}
        return probabilities

    def plot_circuit_characteristics(self,circuit: QuantumCircuit) -> None:
        """
        Creates visual representations of quantum circuit characteristics using multiple subplots.

        Args:
            circuit (QuantumCircuit): The quantum circuit to analyze and visualize

        Returns:
            None: Displays multiple matplotlib figures showing circuit characteristics
        """
        import matplotlib.pyplot as plt
        import numpy as np
        from qiskit.visualization import plot_histogram

        # Create a figure with multiple subplots
        fig = plt.figure(figsize=(15, 10))

        # 1. Circuit Diagram
        plt.subplot(221)
        circuit.draw(output='mpl')
        plt.title('Circuit Diagram')

        # 2. Gate Distribution
        plt.subplot(222)
        gate_counts = circuit.count_ops()
        gates = list(gate_counts.keys())
        counts = list(gate_counts.values())
        plt.bar(gates, counts)
        plt.title('Gate Distribution')
        plt.xticks(rotation=45)

        # 3. Circuit Depth Analysis
        plt.subplot(223)
        depth = circuit.depth() # Use depth() method
        width = circuit.width()
        plt.bar(['Depth', 'Width'], [depth, width])
        plt.title('Circuit Metrics')

        # 4. Qubit Connectivity
        plt.subplot(224)
        num_qubits = circuit.num_qubits
        connectivity = np.zeros((num_qubits, num_qubits))

        for op, qubits, _ in circuit.data:
            if len(qubits) == 2:
                q1, q2 = qubits
                connectivity[q1.index][q2.index] = 1
                connectivity[q2.index][q1.index] = 1

        plt.imshow(connectivity, cmap='Blues')
        plt.colorbar()
        plt.title('Qubit Connectivity')

        plt.tight_layout()
        plt.show()
```

Quantum Boundary Attack,  
\begin{equation*}
\begin{aligned}
x^{\text{adv}} &= \text{clip}_{[0,1]}\left(x^{\text{base}} + \omega\mathcal{N}_q\right) \\
x^{\text{base}} &= \arg\min_{x'} \|\nabla L(f_{\theta}(x'), y)\|_2 \\
\mathcal{N}_q &= \mathcal{Q}_{\omega}(x \in [0,1]^n)
\end{aligned}
\end{equation*}



Quantum Basic Iterative Method,  
\begin{equation*}
\begin{aligned}
x^{k+1} &= \text{clip}_{[0,1]}(x^k + \alpha\text{sign}(g^k)) \\
g^k &= \nabla L(f_{\theta}(x^k), y) + \omega\mathcal{N}_q \\
\mathcal{N}_q &= \mathcal{Q}_{\omega}(x \in [0,1]^n)
\end{aligned}
\end{equation*}





Quantum NewtonFool,  
\begin{equation*}
\begin{aligned}
H(x) &= \frac{\partial^2 L(f_{\theta}(x), y)}{\partial x^2} \\
g(x) &= \nabla L(f_{\theta}(x), y) + \omega\mathcal{N}_q \\
x^{k+1} &= x^k - \eta(H^{-1}g)^k
\end{aligned}
\end{equation*}


 Quantum HopSkipJump,  
\begin{equation*}
\begin{aligned}
x^{k+1} &= x^k + \beta_k\text{sign}(g^k) \\
g^k &= \nabla L(f_{\theta}(x^k), y) + \omega\mathcal{N}_q \\
\beta_k &= \min(\beta_{\max}, \beta_0 + k\Delta\beta)
\end{aligned}
\end{equation*}



 Quantum Noise Injection,  
\begin{equation*}
\begin{aligned}
\mathcal{N}_q &= \mathcal{Q}_{\omega}(x) \\
&= \text{reshape}(\mathcal{Q}'(q=n), (1,1) + \text{shape}(x))
\end{aligned}
\end{equation*}

```python
# @title


class QuantumBoundaryAttack(BoundaryAttack):
    def __init__(self, estimator, targeted=False, max_iter=100,
                 quantum_weight=0.5):
        super().__init__(
            estimator=estimator,
            targeted=targeted,
            max_iter=max_iter

        )
        self.quantum_weight = quantum_weight
        self.qng = QuantumNoiseGenerator(num_qubits=8)

    def generate(self, x, y=None):
        """Generate Boundary attacks with quantum-inspired perturbations."""
        import tensorflow as tf

        # Get initial point
        x_adv = super().generate(x, y)

        # Add quantum noise to the boundary points
        quantum_noise = self.qng.generate_quantum_noise(x.shape[1:])
        # Reshape quantum noise for broadcasting with boundary points
        quantum_noise_reshaped = np.reshape(quantum_noise, (1,) + quantum_noise.shape)
        x_adv = tf.clip_by_value(
            x_adv + self.quantum_weight * quantum_noise_reshaped,
            clip_value_min=0,
            clip_value_max=1
        )

        return x_adv



class QuantumBasicIterativeMethod(BasicIterativeMethod):
    def __init__(self, estimator, eps=0.3, eps_step=0.1,
                 max_iter=40, targeted=False,
                 quantum_weight=0.5):
        super().__init__(
            estimator=estimator,
            eps=eps,
            eps_step=eps_step,
            max_iter=max_iter,
            targeted=targeted

        )
        self.quantum_weight = quantum_weight
        self.qng = QuantumNoiseGenerator(num_qubits=8)

    def generate(self, x, y=None):
        """Generate BIM attacks with quantum-inspired perturbations."""
        import tensorflow as tf

        # Initialize best result
        best_adv = tf.identity(x)

        # Try multiple random initializations
        for i in range(self.num_random_init):
            # Generate random initialization within epsilon ball
            if self.targeted:
                random_perturbation = tf.random.uniform(tf.shape(x),
                                                      minval=-self.eps,
                                                      maxval=self.eps)
            else:
                random_perturbation = tf.random.uniform(tf.shape(x),
                                                      minval=-self.eps,
                                                      maxval=self.eps)
            current = tf.clip_by_value(x + random_perturbation, 0, 1)

            # Perform iterative attack with quantum noise
            for _ in range(self.max_iter):
                # Get gradient with quantum noise injection
                grad = self.estimator.class_gradient(current, y=y)
                quantum_noise = self.qng.generate_quantum_noise(x.shape[1:])
                # Reshape quantum_noise to match gradient shape for broadcasting
                quantum_noise_reshaped = np.reshape(quantum_noise, (1, 1) + quantum_noise.shape)
                noisy_grad = grad + self.quantum_weight * tf.convert_to_tensor(quantum_noise_reshaped)

                # Update adversarial example
                if not self.targeted:
                    current = tf.clip_by_value(
                        current + self.eps_step * tf.sign(noisy_grad),
                        clip_value_min=0,
                        clip_value_max=1
                    )
                else:
                    current = tf.clip_by_value(
                        current - self.eps_step * tf.sign(noisy_grad),
                        clip_value_min=0,
                        clip_value_max=1
                    )

                # Project back onto feasible region
                if not self.targeted:
                    current = tf.clip_by_value(
                        current,
                        clip_value_min=x - self.eps,
                        clip_value_max=x + self.eps
                    )
                else:
                    current = tf.clip_by_value(
                        current,
                        clip_value_min=x - self.eps,
                        clip_value_max=x + self.eps
                    )

            # Update best result
            if self._compare(x, best_adv, current):
                best_adv = current

        return best_adv


class QuantumNewtonFool(NewtonFool):
    def __init__(self, classifier, max_iter=100, eta=0.03,
                 clip_min=None, clip_max=None, batch_size=128,
                 verbose=False, quantum_weight=0.5):
        super().__init__(
            classifier=classifier,
            max_iter=max_iter,
            eta=eta,
            batch_size=batch_size,
            verbose=verbose
        )
        self.quantum_weight = quantum_weight
        self.qng = QuantumNoiseGenerator(num_qubits=8)

    def _compute_gradient_hessian(self, x, y):
        """Compute gradient and Hessian with quantum noise injection."""
        import tensorflow as tf

        # Compute gradient
        grad = self.estimator.class_gradient(x, y=y)

        # Add quantum noise to gradient
        quantum_noise = self.qng.generate_quantum_noise(x.shape[1:])
        # Reshape quantum_noise to match grad shape for broadcasting
        quantum_noise_reshaped = np.reshape(quantum_noise, (1, 1) + quantum_noise.shape)
        noisy_grad = grad + self.quantum_weight * tf.convert_to_tensor(quantum_noise_reshaped)

        # Compute Hessian-vector product
        def hvp(v):
            # Hessian-vector product also needs target labels
            return self.estimator.hessian_vector_product(x, v, y=y)

        return noisy_grad, hvp


class QuantumHopSkipJump(HopSkipJump):
    def __init__(self, classifier, targeted=False, max_iter=64,
                 max_eval=10000, init_eval=100, init_size=32,
                 batch_size=128, quantum_weight=0.5):
        super().__init__(
            classifier=classifier,
            targeted=targeted,
            max_iter=max_iter,
            max_eval=max_eval,
            init_eval=init_eval,
            init_size=init_size,
            batch_size=batch_size
        )
        self.quantum_weight = quantum_weight
        self.qng = QuantumNoiseGenerator(num_qubits=8)

    def _compute_gradient(self, x, y):
        """Compute gradient with quantum noise injection."""
        grad = super()._compute_gradient(x, y)
        quantum_noise = self.qng.generate_quantum_noise(x.shape[1:])
        # Reshape quantum_noise to match grad shape for broadcasting
        quantum_noise_reshaped = np.reshape(quantum_noise, (1, 1) + quantum_noise.shape)
        import tensorflow as tf
        return grad + self.quantum_weight * tf.convert_to_tensor(quantum_noise_reshaped)

def evaluate_all_attacks(model, x_test, y_test):
    """Evaluate all three attack methods with quantum integration."""
    classifier = KerasClassifier(model=model)

    # Initialize attacks
    qba = QuantumBoundaryAttack(estimator=classifier, max_iter=25)
    qhsj = QuantumHopSkipJump(classifier=classifier, max_iter=25)
    qi_fgsm = FastGradientMethod(classifier, eps=0.5)
    qng = QuantumNoiseGenerator(num_qubits=8)

    print("\nAttack Evaluation Results:")

    # Clean accuracy
    loss, accuracy = model.evaluate(x_test[:100], y_test[:100], verbose=0)
    print(f"Clean accuracy: {accuracy:.2f}")

    # Quantum-Inspired FGSM
    fgsm_attacks = []
    for img in x_test[:100]:
        adv_img = qi_fgsm.generate(x=img[np.newaxis])[0]
        quantum_noise = qng.generate_quantum_noise(img.shape)
        final_attack = adv_img + 0.5 * 0.3 * quantum_noise
        final_attack = np.clip(final_attack, 0, 1)
        fgsm_attacks.append(final_attack)


    loss, accuracy = model.evaluate(np.array(fgsm_attacks), y_test[:100], verbose=0)
    print(f"Quantum-Inspired FGSM accuracy: {accuracy:.2f}")

    # Quantum-Inspired HopSkipJump
    hsja_attacks = qhsj.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(hsja_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired HopSkipJump accuracy: {accuracy:.2f}")

    # Quantum-Inspired Boundary Attack
    ba_attacks = qba.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(ba_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired Boundary accuracy: {accuracy:.2f}")


    # Quantum-Inspired NewtonFool
    qnf = QuantumNewtonFool(classifier=classifier, max_iter=10, eta=0.03)
    nf_attacks = qnf.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(nf_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired NewtonFool accuracy: {accuracy:.2f}")


    # Quantum-Inspired BIM
    qbim = QuantumBasicIterativeMethod(estimator=classifier, eps=0.3, max_iter=40)
    bim_attacks = qbim.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(bim_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired BIM accuracy: {accuracy:.2f}")
```

```python
# @title

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.gridspec import GridSpec
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from art.estimators.classification import KerasClassifier
from art.attacks.evasion import FastGradientMethod, BoundaryAttack # Ensure these are imported correctly
from tensorflow.keras.datasets import mnist # Import mnist to load data
from tensorflow.keras.models import Sequential, Model # Import Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, Activation, Add, Input # Import necessary layers
from tensorflow.keras.optimizers import Nadam # Import Nadam optimizer
from tensorflow.keras import regularizers # Import regularizers
from tensorflow.keras import backend as K # Import Keras backend


def create_model():
    def residual_block(x, filters, kernel_size=(3, 3)):
        # Store the input to the block for the shortcut connection
        shortcut = x

        y = Conv2D(filters, kernel_size, padding='same')(x)
        y = BatchNormalization()(y)
        y = Activation('relu')(y)
        y = Conv2D(filters, kernel_size, padding='same')(y)
        y = BatchNormalization()(y)

        # Match the number of filters and spatial dimensions for the shortcut connection
        if K.int_shape(shortcut)[-1] != filters or K.int_shape(shortcut)[1:3] != K.int_shape(y)[1:3]:
             # Use a 1x1 convolution with appropriate stride to match shapes
             # Determine stride based on spatial dimension change
             stride = 1
             if K.int_shape(shortcut)[1] != K.int_shape(y)[1]:
                  stride = 2 # AssumingMaxPooling2D reduced spatial dimensions by a factor of 2

             shortcut = Conv2D(filters, (1, 1), strides=(stride, stride), padding='same')(shortcut)
             shortcut = BatchNormalization()(shortcut) # Optional: Add batch norm to the shortcut


        y = Add()([y, shortcut]) # Add the output to the modified shortcut
        return Activation('relu')(y)

    # Use functional API for residual blocks
    x_input = Input(shape=(28, 28, 1))
    x = Conv2D(32, (3, 3), padding='same')(x_input)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)

    # Residual Blocks
    x = residual_block(x, 32)
    x = MaxPooling2D((2, 2))(x)
    x = Dropout(0.25)(x)

    # Feature Extraction
    # The input x to this block will have 32 filters and reduced spatial dimensions due to MaxPooling2D
    # The block will output layers with 64 filters
    x = residual_block(x, 64)
    x = MaxPooling2D((2, 2))(x)
    x = Dropout(0.25)(x)

    # Dense Layers with Attention
    x = Flatten()(x)
    x = Dense(128, kernel_regularizer=regularizers.L1(0.01))(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Dropout(0.5)(x)

    # Output
    outputs = Dense(10, activation='softmax')(x)

    model = Model(inputs=x_input, outputs=outputs)

    # Advanced Optimization
    optimizer = Nadam(learning_rate=0.002) # Removed schedule_decay
    model.compile(loss='sparse_categorical_crossentropy',
                  optimizer=optimizer,
                  metrics=['accuracy']) # Removed 'top_k_categorical_accuracy'
    return model

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from matplotlib.gridspec import GridSpec
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

def create_visualization(model, x_test, y_test, fgsm_attacks, ba_attacks,
                         hsja_attacks, bim_attacks, nf_attacks):
    """
    Creates a comprehensive visualization of multiple adversarial attack results

    Parameters:
    - model: Keras model
    - x_test: Original test images
    - y_test: Test labels
    - fgsm_attacks: FGSM generated attacks
    - ba_attacks: Boundary attacks
    - hsja_attacks: HopSkipJump attacks
    - bim_attacks: Basic Iterative Method attacks
    - nf_attacks: NewtonFool attacks
    """
    # Set up the style with enhanced aesthetics
    plt.style.use('seaborn-v0_8-darkgrid')
    plt.rcParams['figure.figsize'] = (25, 20)
    plt.rcParams['font.size'] = 10
    plt.rcParams['axes.titlesize'] = 12
    plt.rcParams['axes.labelsize'] = 10

    # Create figure with GridSpec for precise layout control
    fig = plt.figure(figsize=(25, 20))
    gs = GridSpec(5, 6, figure=fig)

    def plot_examples(ax, original, attacked, title):
        """Helper function to plot individual attack examples"""
        ax.imshow(original.reshape(28, 28), cmap='gray')

        # Enhanced prediction display with confidence
        if isinstance(model, Model):
            pred = model.predict(original[np.newaxis,...])
            pred_class = pred.argmax()
            conf = f"{pred.max():.2%}"
            ax.set_title(f'Original\nPred: {pred_class}\nConf: {conf}')
        else:
            ax.set_title('Original\nModel not available for prediction')

        # Add difference inset with improved contrast
        axins = inset_axes(ax, width="35%", height="35%", loc='upper right')
        if original.shape == attacked.shape:
            diff = np.abs(original - attacked).reshape(28, 28)
            im = axins.imshow(diff, cmap='viridis', interpolation='bilinear')
            axins.set_xticks([])
            axins.set_yticks([])
            cbar = plt.colorbar(im, ax=axins, orientation='horizontal', pad=0.1)
            cbar.ax.tick_params(labelsize=8)
        else:
            axins.text(0.5, 0.5, 'Shape Mismatch',
                      horizontalalignment='center',
                      verticalalignment='center',
                      transform=axins.transAxes,
                      color='red')

    def plot_attack_statistics(ax, clean_data, attack_data, title):
        """Helper function to plot statistical comparison"""
        metrics = {
            'Mean Difference': np.mean(np.abs(clean_data - attack_data)),
            'Max Difference': np.max(np.abs(clean_data - attack_data)),
            'Std Dev Diff': np.std(np.abs(clean_data - attack_data))
        }

        sns.barplot(x=list(metrics.keys()),
                   y=list(metrics.values()),
                   palette='Set2',
                   ax=ax)
        ax.set_ylabel('Magnitude')
        ax.set_title(title)
        plt.xticks(rotation=45, ha='right')

    def plot_feature_importance(ax, clean_data, attack_data, title):
        """Helper function to plot feature importance"""
        diff = np.abs(clean_data - attack_data)
        feat_imp = np.mean(diff, axis=0)

        sns.barplot(x=np.arange(len(feat_imp)),
                   y=feat_imp,
                   palette='viridis',
                   ax=ax)
        ax.set_ylabel('Feature Importance')
        ax.set_xlabel('Pixel Index')
        ax.set_title(title)
        plt.xticks([])

    # Plot example comparisons with enhanced spacing
    if len(x_test) > 0:
        plot_examples(fig.add_subplot(gs[0, 0]), x_test[0], fgsm_attacks[0], 'FGSM Attack')
        plot_examples(fig.add_subplot(gs[0, 2]), x_test[0], ba_attacks[0], 'Boundary Attack')
        plot_examples(fig.add_subplot(gs[0, 4]), x_test[0], hsja_attacks[0], 'HopSkipJump Attack')
        plot_examples(fig.add_subplot(gs[0, 5]), x_test[0], bim_attacks[0], 'BIM Attack')
        plot_examples(fig.add_subplot(gs[0, 1]), x_test[0], nf_attacks[0], 'NewtonFool')  # Fixed overlap

    # Enhanced accuracy comparison with statistical measures
    ax_acc = fig.add_subplot(gs[1, :3])
    acc_data = {}
    conf_data = {}

    if isinstance(model, Model):
        # Evaluate on clean data
        clean_eval = model.evaluate(x_test[:100], y_test[:100], verbose=0)
        acc_data['Clean'] = clean_eval[1]
        conf_data['Clean'] = clean_eval[0]

        # Evaluate all attack types
        attack_types = [
            ('FGSM', fgsm_attacks),
            ('Ba', ba_attacks),
            ('BIM', bim_attacks),
            ('NewtonFool', nf_attacks),
            ('HopSkipJump', hsja_attacks)
        ]

        for attack_name, attacks in attack_types:
            if len(attacks) > 0 and attacks.shape == x_test[:100].shape:
                eval_result = model.evaluate(attacks, y_test[:100], verbose=0)
                acc_data[attack_name] = eval_result[1]
                conf_data[attack_name] = eval_result[0]
            else:
                acc_data[attack_name] = 0.0
                conf_data[attack_name] = float('inf')

    # Create enhanced accuracy plot with error bars
    attack_names = list(acc_data.keys())
    accuracies = [acc_data[name] for name in attack_names]
    losses = [conf_data[name] for name in attack_names]

    x = np.arange(len(accuracies))
    width = 0.35

    ax_acc.bar(x - width/2, accuracies, width, label='Accuracy', color='skyblue')
    ax_acc.bar(x + width/2, losses, width, label='Loss', color='lightcoral')

    ax_acc.set_ylabel('Value')
    ax_acc.set_title('Attack Performance Comparison')
    ax_acc.set_xticks(x)
    ax_acc.set_xticklabels(attack_names, rotation=45, ha='right')
    ax_acc.legend()

    # Plot statistical comparisons
    plot_attack_statistics(fig.add_subplot(gs[1, 3]), x_test[:100], fgsm_attacks[:100], 'FGSM Statistics')
    plot_attack_statistics(fig.add_subplot(gs[1, 4]), x_test[:100], ba_attacks[:100], 'Boundary Attack Statistics')
    plot_attack_statistics(fig.add_subplot(gs[1, 5]), x_test[:100], hsja_attacks[:100], 'HopSkipJump Statistics')

    # Plot feature importance analysis
    def plot_feature_importance(ax, clean_data, attack_data, title):
        """
        Helper function to plot feature importance analysis between clean and attacked data

        Parameters:
        - ax: matplotlib axes object
        - clean_data: Original clean data
        - attack_data: Attacked data
        - title: Title for the plot
        """
        # Calculate feature importance using permutation importance
        diff = np.abs(clean_data - attack_data)
        feat_imp = np.mean(diff, axis=0)

        # Create DataFrame for better visualization
        import pandas as pd
        df = pd.DataFrame({
           'Feature': range(len(feat_imp)),
           'Importance': feat_imp
                 })

        # Sort by importance
        df = df.sort_values('Importance', ascending=True)

        # Create horizontal bar plot
        sns.barplot(x='Importance', y='Feature',
                data=df,
                color='skyblue',
                ax=ax)

        # Customize the plot
        ax.set_title(title)
        ax.set_xlabel('Feature Importance')
        ax.set_ylabel('Feature Index')

        # Add value labels
        for i, v in enumerate(df['Importance']):
           ax.text(v, i, f'{v:.3f}', va='center')

        # Adjust layout
        plt.tight_layout()

        # Add to the existing GridSpec layout
        plot_feature_importance(fig.add_subplot(gs[2, 3:]), x_test[:100], fgsm_attacks[:100], 'FGSM Feature Importance')
        plot_feature_importance(fig.add_subplot(gs[3, 3:]), x_test[:100], ba_attacks[:100], 'Boundary Attack Feature Importance')
        plot_feature_importance(fig.add_subplot(gs[4, 3:]), x_test[:100], hsja_attacks[:100], 'HopSkipJump Feature Importance')


def evaluate_and_visualize():
    """Evaluates attacks and creates visualization"""
    # Load MNIST dataset
    (x_train, y_train), (x_test, y_test) = mnist.load_data()
    x_train = x_train.astype('float32') / 255
    x_test = x_test.astype('float32') / 255
    x_train = x_train.reshape(-1, 28, 28, 1)
    x_test = x_test.reshape(-1, 28, 28, 1)

    # Create and train model
    model = create_model()
    model.fit(x_train, y_train, batch_size=128, epochs=1, verbose=1)

    # Generate attacks
    classifier = KerasClassifier(model=model)
    qba = QuantumBoundaryAttack(estimator=classifier, max_iter=30)
    qhsj = QuantumHopSkipJump(classifier=classifier, max_iter=30)
    qi_fgsm = FastGradientMethod(classifier, eps=0.5)
    qng = QuantumNoiseGenerator(num_qubits=8)


    # Quantum-Inspired FGSM
    fgsm_attacks = []
    for img in x_test[:100]:
        adv_img = qi_fgsm.generate(x=img[np.newaxis])[0]
        quantum_noise = qng.generate_quantum_noise(img.shape)
        final_attack = adv_img + 0.5 * 0.3 * quantum_noise
        final_attack = np.clip(final_attack, 0, 1)
        fgsm_attacks.append(final_attack)

    fgsm_attacks = np.array(fgsm_attacks) # Convert list to numpy array


    # Quantum-Inspired HopSkipJump
    hsja_attacks = qhsj.generate(x=x_test[:100], y=y_test[:100],max_iter=30)
    loss, accuracy = model.evaluate(hsja_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired HopSkipJump accuracy: {accuracy:.2f}")

    # Quantum-Inspired Boundary Attack
    ba_attacks = qba.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(ba_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired Boundary accuracy: {accuracy:.2f}")


    # Quantum-Inspired NewtonFool
    qnf = QuantumNewtonFool(classifier=classifier, max_iter=30, eta=0.5)
    nf_attacks = qnf.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(nf_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired NewtonFool accuracy: {accuracy:.2f}")


    # Quantum-Inspired BIM
    qbim = QuantumBasicIterativeMethod(estimator=classifier, eps=0.5, max_iter=40)
    bim_attacks = qbim.generate(x=x_test[:100], y=y_test[:100])
    loss, accuracy = model.evaluate(bim_attacks, y_test[:100], verbose=0)
    print(f"Quantum-Inspired BIM accuracy: {accuracy:.2f}")


    # Create visualization
    fig = create_visualization(
        model, x_test[:100], y_test[:100],
        fgsm_attacks, ba_attacks,
        hsja_attacks, bim_attacks, nf_attacks
    )
    plt.savefig('attack_analysis.png', dpi=300, bbox_inches='tight')
    print("\nDetailed Evaluation Metrics:")
    print("-" * 50)


def calculate_metrics(original, attacked):
     """Helper function to calculate attack metrics"""
     if original.shape != attacked.shape or original.size == 0:
        return {
                'Mean Perturbation': np.nan,
                'Max Perturbation': np.nan,
                'Std Perturbation': np.nan,
                'Median Perturbation': np.nan
            }
     diff = np.abs(original - attacked)
     return {
         'Mean Perturbation': np.mean(diff),
         'Max Perturbation': np.max(diff),
         'Std Perturbation': np.median(diff), # Corrected typo: changed np.std to np.median
         'Median Perturbation': np.median(diff)
     }

evaluate_and_visualize()
```
